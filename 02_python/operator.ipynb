```python
# 連続比較の例
x = 150
print("100 < x < 200: ", 100 < x < 200)  # True

y = 50
print("100 < y < 200: ", 100 < y < 200)  # False

# 論理演算子の例
x = 15
print("10 < x and x < 20: ", 10 < x and x < 20)  # True
print("x < 10 or x > 20: ", x < 10 or x > 20)    # False
print("not (x == 15): ", not (x == 15))          # False
```
