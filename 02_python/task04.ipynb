```python
try:
    dividend = int(input("割られる数を入力してください: "))
    divisor = int(input("割る数を入力してください: "))
    if divisor == 0:
        raise ZeroDivisionError("エラー: 0 で割ることはできません")
    result = dividend / divisor
except ValueError:
    print("エラー: 数値を入力してください")
except ZeroDivisionError as e:
    print(e)
else:
    print(f"結果: {result}")
```
