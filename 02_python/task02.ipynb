```python
TAX_RATE = 10 # 消費税
SHIP_COST = 350 # 送料
SHIP_FREE_THRESHOLD = 2000 # 送料が無料になる価格

price = int(input("商品の価格を入力してください: "))

tax = price * TAX_RATE // 100
price_with_tax = price + tax

if price_with_tax < SHIP_FREE_THRESHOLD:
    shipping = SHIP_COST
else:
    shipping = 0

total_price = price_with_tax + shipping

if shipping == 0:
    print("送料は無料です。")
else:
    print(f"送料は{shipping}円です。")

print(f"送料込税込価格は{total_price}円です。")
```
