```python
# バリデーション関数
def validate(text: str) -> bool:
    if len(text) < 5 or len(text) > 15:
        print("エラー: ユーザー名は5文字以上15文字以内で入力してください")
        return False

    if " " in text:
        print("エラー: ユーザー名に空白を含めることはできません")
        return False

    if not text.isalnum():
        print("エラー: ユーザー名は英数字のみ使用できます")
        return False
    return True 

# main関数
def main():
    username = input("ユーザー名を入力してください: ")
    if validate(username):
        print("ユーザー名は有効です")
    else:
        print("不正なユーザー名です")
        
main()
```
