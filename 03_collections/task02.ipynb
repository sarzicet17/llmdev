```python
my_dict = {'a':0, 'b':0, 'c':0, 'd':0, 'e':0, 'f':0, 'g':0, 'h':0, 'i':0, 'j':0, 'k':0, 'l':0, 'm':0, 'n':0, 'o':0, 'p':0, 'q':0, 'r':0, 's':0, 't':0, 'u':0, 'v':0, 'w':0, 'x':0, 'y':0, 'z':0}

items = []

# ユーザーからの入力を受け取る。""で終了
# リストに追加していく
while True:
    user_input = input("英単語を入力してください: ")
    if user_input == "":
        break
    items.append(user_input)
items.sort()
print("入力した英単語:", items)

# リスト内の各単語の文字を辞書のキーと照合し、出現回数をカウントし出力
for item in items:
    for char in item:
        if char in my_dict:
            my_dict[char] += 1
for key, value in my_dict.items():
    if value > 0:
        print(f"{key}が{value}個ありました")
            
```
