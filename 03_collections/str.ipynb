```python
s = "abcdefghijklmnopqrstuvwxyz"

print(s[:10]) # 開始位置を省略
print(s[10:]) # 終了位置を省略
print(s[:])   # 両方省略

# vowelのみ
s = "TechAcademy"
vowels = "aeiouAEIOU"  # 母音のリスト
for char in s:
    if char in vowels:
        print(char)
```
